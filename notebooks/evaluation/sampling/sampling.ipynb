```python
import polars as pl

file_name = "results_EN_NPO_e20-ue3"
file_path = f"../../../results/extraction/{file_name}.parquet"

df = pl.read_parquet(file_path)
RANDOM_SEED = 42
n = 20
```

```python
forget_df = df.filter(pl.col('is_forget') == True)
retain_df = df.filter(pl.col('is_forget') == False)
```

```python
forget_df = forget_df.unique(subset=["expected_answer"]).sample(n=n, seed=RANDOM_SEED)
retain_df = retain_df.unique(subset=["expected_answer"]).sample(n=n, seed=RANDOM_SEED)
```

```python
combined_df = pl.concat([
    forget_df,
    retain_df
])
```

```python
combined_df = combined_df.sample(fraction=1.0, shuffle=True, seed=RANDOM_SEED)
```

```python
combined_df.write_parquet(f"../../../results/extraction/{file_name}-subset.parquet")
```
