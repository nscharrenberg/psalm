```python
from dotenv import load_dotenv
from deepeval.metrics import DAGMetric, DeepAcyclicGraph
from deepeval.metrics.dag import TaskNode, VerdictNode, NonBinaryJudgementNode
from deepeval.test_case import LLMTestCaseParams

from psalm.evaluators.llm.base_dag_evaluator import BaseDagEvaluator
from psalm.evaluators.llm.narrative import CharacterSimilarityEvaluator

load_dotenv()
```

```python
from psalm.core.models import Book, AnonymousText

evaluator = CharacterSimilarityEvaluator()

source_text = Book(
    title="Tides of Memory",
    author="Noah Scharrenberg",
    text="Mara wore a coat with a moth-badge that glowed when the harbor fog thickened. She kept a jar of sea-salt on her shelf, a ritual to remind her that stories could be saved from the waves. She counted thirteen bakery steps to steady her nerves, like a quiet drumbeat in the morning. Her left eyebrow twitched when she lied, a faint metronome she could not hush. Mara wrote letters to a father who drifted away, never sent, and spoke to the moon as if it were a friend. At night she traced the pattern on a blue scarf, turning memories into careful maps. The town called her quiet, but inside, storms brewed, weaving meaning from small, ordinary acts.",
    language="English"
)

target_text_9 = AnonymousText(
    text="Mara wore a coat with a moth-badge that glowed when the harbor fog thickened. She kept a jar of sea-salt on her shelf, a ritual to remind her that stories could be saved from the waves. She counted thirteen bakery steps to steady her nerves, like a quiet drumbeat in the morning. Her left eyebrow twitched when she lied, a faint metronome she could not hush. Mara wrote letters to a father who drifted away, never sent, and spoke to the moon as if it were a friend. At night she traced the pattern on a blue scarf, turning memories into careful maps. The town called her quiet, but inside, storms brewed, weaving meaning from small, ordinary acts."
)

result, metric_result = evaluator.evaluate_with_metric(source=source_text, target=target_text_9, debug=True, model_name="gpt-5-nano")
```
