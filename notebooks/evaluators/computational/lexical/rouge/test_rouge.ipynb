```python
from psalm.core.models import Book, AnonymousText
from psalm.evaluators.computational.lexical import RougeEvaluator, BleuEvaluator

source = Book(text="""
But I can't take it any longer. I want to leave them and marry. Then I will be my own master, because my husband, he will make me dance to my tune. I already have three suitors; but the men, I have heard from my grandmother, are a false race, many beautiful promises, but without a word. You have to keep them well in sight, and I will do that with the three who come to court me. The sweeper Spiegels is a good and charming man, he has some fortune, but he is not handsome; you will tell me that he sits on his wife's stool, earns a lot of money and sweeps a lot, that is true, but.... he is not handsome at all. The clapperman Moentjens is a bit more handsome, but he has been at college for a day and a morning and speaks to me too sweetly; besides, he walks around too much at night on the street, I have no trust in that. Only the little Pierssens, who plays the trumpet at the Harmony, he does please me; but would he want to propose to me at the town hall?
""", title="Het kindermaögske, klúchtspeul in ein akt (onder ps. ene boer en twie lui)", author=" A. Loisel", language="English")

target = AnonymousText(text="""
Antje.  Oh! those two suitors! One is so fat, I am already married to him; the other one is so thin, I am still waiting for him with buns. (Lies in the arms of
the fat man.) Hey! hey! I am already pregnant! (To the thin man.) Come, man, you must know that you are the only one left, I am giving up on the fat one.
""")

rouge_evaluator = RougeEvaluator()
rouge_result = rouge_evaluator.evaluate(source, target)
```

```python
print(f"ROUGE Score: {rouge_result.score}")
```

```python
bleu_evaluator = BleuEvaluator()
bleu_result = bleu_evaluator.evaluate(source, target)
```

```python
print(f"BLEU Score: {bleu_result.score}")
```
