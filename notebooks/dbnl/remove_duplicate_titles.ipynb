```python
import os
import polars as pl

GENRE = "proza"
DATASET_PATH = f"data/splitted_1850_text_length_max_70k_top_5_most_published/{GENRE}"
AUTHORS = [
    "_kle007",
    "_kon002",
    "goej001",
    "goev001",
    "pier003"
]

ti_ids_used = []
for author in AUTHORS:
    file_path = os.path.join(DATASET_PATH, f"{author}.parquet")
    temp_df = pl.read_parquet(file_path)

    indexes_to_remove = []

    for row in temp_df.iter_rows(named=True):
        ti_id = row["ti_id"]

        if ti_id in ti_ids_used:
            indexes_to_remove.append(ti_id)
        else:
            ti_ids_used.append(ti_id)

    temp_df.remove(pl.col("ti_id").is_in(indexes_to_remove))

    temp_df.write_parquet(os.path.join(DATASET_PATH, f"{author}.parquet"))
```
