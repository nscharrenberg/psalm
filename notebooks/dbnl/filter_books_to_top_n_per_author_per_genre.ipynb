```python
from datasets import load_dataset
import polars as pl

DATASET_NAME = "nscharrenberg/DBNL-public"
DATASET_SUBSET = "tabular_original"
DATASET_SPLIT = "train"

ds = load_dataset(DATASET_NAME, DATASET_SUBSET, split=DATASET_SPLIT)
```

```python
from datasets import Dataset

df = Dataset.to_polars(ds)
```

```python
df["genre"].unique()
```

```python
from dbnl.filters.filter_by_year import filter_by_year

year_df = filter_by_year(df, 1850, verbose=True)
```

```python
from dbnl.filters.filter_by_genre import filter_by_genres_separated

genre_dfs = filter_by_genres_separated(year_df, ["jeugdliteratuur", "poëzie", "proza", "drama"], verbose=True)
```

```python
from dbnl.filters.filter_by_author import filter_by_authors_separated

genre_author_grouped_dfs: dict[str, dict[str, pl.DataFrame]] = {}

for genre, genre_df in genre_dfs.items():
    authors = genre_df["pers_id"].unique().to_list()
    genre_author_grouped_dfs[genre] = filter_by_authors_separated(genre_df, authors, verbose=True)
```

```python
import os

SAVE_PATH = "data/splitted_1850"
for genre, author_grouped_df in genre_author_grouped_dfs.items():
    genre_path = os.path.join(SAVE_PATH, genre)
    os.makedirs(genre_path, exist_ok=True)
    for author, author_df in author_grouped_df.items():
        file_name = f"{author}.parquet"
        author_path = os.path.join(genre_path, file_name)
        author_df.write_parquet(author_path)
```

```python
top_n_authors_per_genre: dict[str, dict[str, int]] = {}
n = 5

for genre, author_grouped_df in genre_author_grouped_dfs.items():
    top_n_authors_per_genre[genre] = {}
    for author, author_df in author_grouped_df.items():
        top_n_authors_per_genre[genre][author] = len(author_df)

    top_n_authors_per_genre[genre] = dict(sorted(top_n_authors_per_genre[genre].items(), key=lambda item: item[1], reverse=True)[:n])
```

```python
top_n_authors_per_genre
```

```python
from dbnl.filters.filter_by_text_length import filter_by_text_length

genre_author_text_limited_dfs: dict[str, dict[str, pl.DataFrame]] = {}
min_length = 300
max_length = 70000

for genre, author_grouped_df in genre_author_grouped_dfs.items():
    genre_author_text_limited_dfs[genre] = {}
    for author, author_df in author_grouped_df.items():
        text_length_filtered_author_df = filter_by_text_length(author_df, min_length, max_length, verbose=False)

        if len(text_length_filtered_author_df) == 0:
            continue

        genre_author_text_limited_dfs[genre][author] = text_length_filtered_author_df
```

```python
top_n_authors_per_genre: dict[str, dict[str, int]] = {}
n = 5

for genre, author_grouped_df in genre_author_text_limited_dfs.items():
    top_n_authors_per_genre[genre] = {}
    for author, author_df in author_grouped_df.items():
        top_n_authors_per_genre[genre][author] = len(author_df)

    top_n_authors_per_genre[genre] = dict(sorted(top_n_authors_per_genre[genre].items(), key=lambda item: item[1], reverse=True)[:n])
```

```python
top_n_authors_per_genre
```

```python
import os

SAVE_PATH = "data/splitted_1850_text_length_max_70k"
for genre, author_grouped_df in genre_author_text_limited_dfs.items():
    genre_path = os.path.join(SAVE_PATH, genre)
    os.makedirs(genre_path, exist_ok=True)
    for author, author_df in author_grouped_df.items():
        file_name = f"{author}.parquet"
        author_path = os.path.join(genre_path, file_name)
        author_df.write_parquet(author_path)
```

```python
import os

SAVE_PATH = "data/splitted_1850_text_length_max_70k_top_5_most_published"
for genre, author_grouped_df in genre_author_text_limited_dfs.items():
    if genre not in top_n_authors_per_genre:
        continue

    genre_path = os.path.join(SAVE_PATH, genre)
    os.makedirs(genre_path, exist_ok=True)

    for author, author_df in author_grouped_df.items():
        if author not in top_n_authors_per_genre[genre]:
            continue

        authors = top_n_authors_per_genre[genre][author]
        file_name = f"{author}.parquet"
        author_path = os.path.join(genre_path, file_name)
        author_df.write_parquet(author_path)
```

```python
total_books_left = 0
for genre, author_grouped_df in top_n_authors_per_genre.items():
    for author, book_count in author_grouped_df.items():
        total_books_left += book_count

print(f"Total books left: {total_books_left}")
```
